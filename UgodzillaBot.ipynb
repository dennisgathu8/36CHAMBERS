```python
import asyncio
from telegram import Update, ForceReply
from telegram.ext import filters, Application, CommandHandler, MessageHandler, CallbackContext
import threading

# Replace 'YOUR_TOKEN' with the API token you received from BotFather
TOKEN = '8149381933:AAE01vEWeAiYQ_5ab-WTjvPvRjjhnuBSwSk'

# Function to handle the '/start' command
async def start(update: Update, context: CallbackContext) -> str:
    await update.message.reply_text('Hello..The ziz-zag-zigla will respond')
    return ""

# Function to handle messages
async def reply_to_mention(update: Update, context: CallbackContext) -> None:
    if 'your_username' in [mention.username for mention in update.message.entities if mention.type == 'mention']:
        reply_text = "Thank you for mentioning me! Your message has been received."
        await update.message.reply_text(reply_text, reply_to_message_id=update.message.message_id, allow_sending_without_reply=True)

async def main() -> None:
    # Create the Application and pass it your bot's token
    application = Application.builder().token(TOKEN).build()

    # Initialize the application (required in v20+)
    await application.initialize()

    # Register the command handler and message handler with the application
    application.add_handler(CommandHandler("start", start))
    application.add_handler(MessageHandler(filters.TEXT, reply_to_mention))

    # Start the Bot
    await application.start()
    await application.updater.start_polling()
    await application.updater.idle()

# Function to start the bot in a separate thread
def run_bot():
    try:
        loop = asyncio.get_running_loop()
    except RuntimeError:
        loop = asyncio.new_event_loop()
        asyncio.set_event_loop(loop)

    loop.run_until_complete(main())

# Running the bot in a separate thread
thread = threading.Thread(target=run_bot)
thread.start()
```
