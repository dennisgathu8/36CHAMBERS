```python
import zipfile
import os
import librosa
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

print("Upload the Victims_of_Madness_2.zipfile")
uploaded = files.upload()

zip_filename = list(uploaded.keys())[0]
with open(zip_filename, "wb") as f:
  f.write(uploaded[zip_filename])

# define extraction path
extract_path = "/content/extracted_audio"
os.makedirs(extract_path, exist_ok=True)
# extract ZIP
with zipfile.ZipFile(zip_filename, "r") as zip_ref:
  zip_ref.extractall(extract_path)
```

```python
# find audio files
audio_files = []
for root, _, files in os.walk(extract_path):
  for f in files:
    if f.endswith((".mp3", ".wav")):
      audio_files.append(os.path.join(root, f))
print("Extracted audio files:", audio_files)

# Extract Comprehensive Audio Features
features_list = []
for audio_file in audio_files:
    try:
        # Load audio
        y, sr = librosa.load(audio_file, sr=None)

        # Temporal Features
        tempo, beat_frames = librosa.beat.tempo(y=y, sr=sr)
        duration = librosa.get_duration(y=y, sr=sr)
        zero_crossing_rate = np.mean(librosa.feature.zero_crossing_rate(y=y))

        # Spectral Features
        spectral_centroid = np.mean(librosa.feature.spectral_centroid(y=y, sr=sr))
        spectral_rolloff = np.mean(librosa.feature.spectral_rolloff(y=y, sr=sr))
        spectral_bandwidth = np.mean(librosa.feature.spectral_bandwidth(y=y, sr=sr))
        spectral_contrast = np.mean(librosa.feature.spectral_contrast(y=y, sr=sr), axis=1)
        chroma = np.mean(librosa.feature.chroma_stft(y=y, sr=sr), axis=1)

        # Rhythmic Features
        onset_env = librosa.onset.onset_strength(y=y, sr=sr)
        onset_strength = np.mean(onset_env)
        beat_strength = np.mean(onset_env[beat_frames]) if beat_frames.size > 0 else 0

        # Timbral Features
        mfcc = np.mean(librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13), axis=1)
        mfcc_delta = np.mean(librosa.feature.delta(mfcc), axis=1)
        mfcc_delta2 = np.mean(librosa.feature.delta(mfcc, order=2), axis=1)

        # Energy Features
        rms_energy = np.mean(librosa.feature.rms(y=y))

        # Pitch Features
        pitches, magnitudes = librosa.piptrack(y=y, sr=sr)
        pitch_mean = np.mean(pitches[pitches > 0]) if np.any(pitches > 0) else 0

        # Store features
        features = {
            "track": os.path.basename(audio_file),
            "tempo_bpm": float(tempo),
            "duration_s": duration,
            "zero_crossing_rate": zero_crossing_rate,
            "spectral_centroid": spectral_centroid,
            "spectral_rolloff": spectral_rolloff,
            "spectral_bandwidth": spectral_bandwidth,
            "spectral_contrast_1": spectral_contrast[0],  # First band for brevity
            "chroma_mean": np.mean(chroma),
            "onset_strength": onset_strength,
            "beat_strength": beat_strength,
            "rms_energy": rms_energy,
            "pitch_mean_hz": pitch_mean
        }
        # Add MFCCs (13 coefficients)
        for i in range(13):
            features[f"mfcc_{i+1}"] = mfcc[i]
            features[f"mfcc_delta_{i+1}"] = mfcc_delta[i]
            features[f"mfcc_delta2_{i+1}"] = mfcc_delta2[i]

        features_list.append(features)
        print(f"Processed: {audio_file}")
    except Exception as e:
        print(f"Error processing {audio_file}: {e}")
```

```python
# Create DataFrame
df = pd.DataFrame(features_list)
df.to_csv("audio_features.csv", index=False)
print("\nAudio Features DataFrame (first 5 rows, selected columns):")
print(df[["track", "tempo_bpm", "rms_energy", "spectral_centroid", "mfcc_1"]].head())
```

```python

# Step 2: Extract Comprehensive Audio Features (limit to 5 tracks to avoid memory issues)
features_list = []
max_tracks = 5  # Adjust this number if needed
for i, audio_file in enumerate(audio_files[:max_tracks]):
    try:
        print(f"Processing track {i+1}/{min(max_tracks, len(audio_files))}: {audio_file}")
        y, sr = librosa.load(audio_file, sr=None)

        # Temporal Features
        tempo = librosa.feature.rhythm.tempo(y=y, sr=sr)[0]  # Updated for librosa 0.10.0+
        duration = librosa.get_duration(y=y, sr=sr)
        zero_crossing_rate = np.mean(librosa.feature.zero_crossing_rate(y=y))

        # Spectral Features
        spectral_centroid = np.mean(librosa.feature.spectral_centroid(y=y, sr=sr))
        spectral_rolloff = np.mean(librosa.feature.spectral_rolloff(y=y, sr=sr))
        spectral_bandwidth = np.mean(librosa.feature.spectral_bandwidth(y=y, sr=sr))
        spectral_contrast = np.mean(librosa.feature.spectral_contrast(y=y, sr=sr), axis=1)
        chroma = np.mean(librosa.feature.chroma_stft(y=y, sr=sr), axis=1)

        # Rhythmic Features
        onset_env = librosa.onset.onset_strength(y=y, sr=sr)
        onset_strength = np.mean(onset_env)
        # Get beat frames for beat strength
        _, beat_frames = librosa.beat.beat_track(y=y, sr=sr)
        beat_strength = np.mean(onset_env[beat_frames]) if beat_frames.size > 0 else 0

        # Timbral Features
        mfcc = np.mean(librosa.feature.mfcc(y=y, sr=sr, n_mfcc=13), axis=1)
        mfcc_delta = np.mean(librosa.feature.delta(mfcc), axis=1)
        mfcc_delta2 = np.mean(librosa.feature.delta(mfcc, order=2), axis=1)

        # Energy Features
        rms_energy = np.mean(librosa.feature.rms(y=y))

        # Pitch Features
        pitches, magnitudes = librosa.piptrack(y=y, sr=sr)
        pitch_mean = np.mean(pitches[pitches > 0]) if np.any(pitches > 0) else 0

        # Store features
        features = {
            "track": os.path.basename(audio_file),
            "tempo_bpm": float(tempo),
            "duration_s": duration,
            "zero_crossing_rate": zero_crossing_rate,
            "spectral_centroid": spectral_centroid,
            "spectral_rolloff": spectral_rolloff,
            "spectral_bandwidth": spectral_bandwidth,
            "spectral_contrast_1": spectral_contrast[0],
            "chroma_mean": np.mean(chroma),
            "onset_strength": onset_strength,
            "beat_strength": beat_strength,
            "rms_energy": rms_energy,
            "pitch_mean_hz": pitch_mean
        }
        for i in range(13):
            features[f"mfcc_{i+1}"] = mfcc[i]
            features[f"mfcc_delta_{i+1}"] = mfcc_delta[i]
            features[f"mfcc_delta2_{i+1}"] = mfcc_delta2[i]

        features_list.append(features)
    except Exception as e:
        print(f"Error processing {audio_file}: {e}")

# Create DataFrame
df = pd.DataFrame(features_list)
if df.empty:
    print("ERROR: No features extracted. Check audio files for compatibility (MP3/WAV) or try re-uploading the ZIP.")
else:
    df.to_csv("audio_features.csv", index=False)
    print("\nAudio Features DataFrame (first 5 rows, selected columns):")
    print(df[["track", "tempo_bpm", "rms_energy", "spectral_centroid", "mfcc_1"]].head())
```

```python
# Step 3: Analyze Features for Hip-Hop Characteristics
print("\nSummary Statistics:")
print(df[["tempo_bpm", "rms_energy", "spectral_centroid", "zero_crossing_rate", "mfcc_1"]].describe())

# Hip-hop tempo range
hip_hop_tempo_range = (80, 110)
print(f"\nTracks in hip-hop tempo range ({hip_hop_tempo_range[0]}-{hip_hop_tempo_range[1]} BPM):")
tempo_matches = df[(df["tempo_bpm"] >= hip_hop_tempo_range[0]) & (df["tempo_bpm"] <= hip_hop_tempo_range[1])]
print(tempo_matches[["track", "tempo_bpm"]])

# High-energy tracks
high_energy_threshold = df["rms_energy"].quantile(0.75)
print(f"\nHigh-energy tracks (RMS energy > {high_energy_threshold:.4f}):")
print(df[df["rms_energy"] > high_energy_threshold][["track", "rms_energy"]])

# Vocal presence (using MFCC_1)
print("\nTracks with high vocal presence (sorted by MFCC_1):")
print(df[["track", "mfcc_1"]].sort_values(by="mfcc_1", ascending=False))

# Bass-heavy tracks (low spectral centroid)
bass_threshold = df["spectral_centroid"].quantile(0.25)
print(f"\nBass-heavy tracks (spectral centroid < {bass_threshold:.1f} Hz):")
print(df[df["spectral_centroid"] < bass_threshold][["track", "spectral_centroid"]])

# Rhythmic intensity (high onset strength)
print("\nTracks with strong rhythmic intensity (sorted by onset strength):")
print(df[["track", "onset_strength"]].sort_values(by="onset_strength", ascending=False))
```

```python
# Step 4: Generate Data-Driven Insights
insights = []
if not df.empty:
    tempo_percentage = len(tempo_matches) / len(df) * 100
    insights.append(f"{tempo_percentage:.1f}% of tracks in Victims of Madness 2 hit the classic hip-hop tempo range (80–110 BPM), ideal for rhythmic flows.")
    high_energy_count = len(df[df["rms_energy"] > high_energy_threshold])
    insights.append(f"{high_energy_count} tracks deliver high-energy beats, perfect for hip-hop anthems.")
    top_vocal_track = df.loc[df["mfcc_1"].idxmax(), "track"]
    insights.append(f"Tracks like {top_vocal_track} showcase strong rap delivery, a hallmark of Wakadinali’s lyrical prowess.")
    bass_count = len(df[df["spectral_centroid"] < bass_threshold])
    insights.append(f"{bass_count} tracks feature bass-heavy production, driving the album’s hip-hop vibe.")
    top_rhythmic_track = df.loc[df["onset_strength"].idxmax(), "track"]
    insights.append(f"Tracks like {top_rhythmic_track} pack intense rhythms, keeping listeners locked in.")

print("\nData-Driven Insights:")
for i, insight in enumerate(insights, 1):
    print(f"{i}. {insight}")
```
