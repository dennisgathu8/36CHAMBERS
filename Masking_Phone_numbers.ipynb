```python
import pandas as pd
from google.colab import files

# Function to mask phone numbers in the required format: 25190xxxx978
def mask_phone_number(phone):
    return phone[:5] + 'xxxx' + phone[-3:]

# Step 1: Upload an Excel file
print("Please upload an Excel file:")
uploaded = files.upload()

# Step 2: Read the uploaded file
for file_name in uploaded.keys():
    df = pd.read_excel(file_name)

# Step 3: Check if 'username' column exists
if 'username' in df.columns:
    # Step 4: Mask the phone numbers in the 'username' column
    df['masked_username'] = df['username'].astype(str).apply(mask_phone_number)
    print("Masked phone numbers added as a new column.")

    # Step 5: Display the dataframe with the masked phone numbers
    print(df.head())

    # Step 6: Save the modified file
    output_file_name = 'Highest Multiplier Promo Winners 29th.xlsx'
    df.to_excel(output_file_name, index=False)
    print(f"File saved as {output_file_name}")

    # Step 7: Download the file
    files.download(output_file_name)
else:
    print("The file does not contain a 'username' column.")
```
