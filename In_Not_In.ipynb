```python
import pandas as pd
from google.colab import files

# Upload the first Excel workbook
print("Please upload the first Excel file")
uploaded1 = files.upload()

# Load the first Excel workbook
file1_name = list(uploaded1.keys())[0]
df1 = pd.read_excel(file1_name)

# Upload the second Excel workbook
print("Please upload the second Excel file")
uploaded2 = files.upload()

# Load the second Excel workbook
file2_name = list(uploaded2.keys())[0]
df2 = pd.read_excel(file2_name)

# Merge the dataframes on 'username' (common records)
merged_df = pd.merge(df1, df2, on='id', how='inner')

# Get the data from df1 with no matching 'username' in df2
non_matching_df = df1[~df1['id'].isin(df2['id'])]

# Save the merged dataframe to a new Excel file
merged_output_path = 'In_Nov_Jan.xlsx'
merged_df.to_excel(merged_output_path, index=False)

# Save the non-matching dataframe to another Excel file
non_matching_output_path = 'In_Nov_not_Jan.xlsx'
non_matching_df.to_excel(non_matching_output_path, index=False)

# Download both files
print("Downloading the merged data file...")
files.download(merged_output_path)

print("Downloading the non-matching data file...")
files.download(non_matching_output_path)

# Display the merged dataframe
print("In Both")
print(merged_df.head())

# Display the non-matching dataframe
print("In Chapa not DB")
print(non_matching_df.head())
```
